```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
url = "https://pubchem.ncbi.nlm.nih.gov/rest/pug/compound/cid/2244,1983,702,6322,5957/property/MolecularWeight,XLogP,InChIKey,CanonicalSMILES/CSV"
df = pd.read_csv(url)
df.head()
```

```python
df.info()
df.isnull().sum()
```

```python
df_clean = df.dropna()
df_clean
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))
plt.scatter(df_clean["MolecularWeight"], df_clean["XLogP"], color='red')
plt.title("Molecular Weight vs XLogP")
plt.xlabel("Molecular Weight")
plt.ylabel("XLogP")
plt.grid(True)
plt.show()
```

```python
plt.bar(df_clean["InChIKey"], df_clean["MolecularWeight"])
plt.title("Molecular Weight of Compounds")
plt.xlabel("Compound")
plt.ylabel("Molecular Weight")
plt.xticks(rotation=20,ha='right')
plt.tight_layout()
plt.show()
```
